```python
import time
import serial

PORT = "COM3"      
BAUD = 115200

ser = serial.Serial(PORT, BAUD, timeout=1)
time.sleep(2)  # tunggu board reset

while ser.in_waiting:
    print(ser.readline().decode(errors="ignore").strip())

def send(cmd: str):
    ser.write((cmd.strip() + "\n").encode())

def read_lines(seconds=0.2):
    time.sleep(seconds)
    out = []
    while ser.in_waiting:
        out.append(ser.readline().decode(errors="ignore").strip())
    return out

# LED ON
send("LED 1")
print("\n".join(read_lines()))

# Fade 100 -> 0
for p in range(100, -1, -10):
    send(f"POWER {p}")
    lines = read_lines(0.15)
    if lines:
        print(lines[-1])

# LED OFF
send("LED 0")
print("\n".join(read_lines()))

ser.close()
print("Disconnected")
```
